```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Objectif

Dans ce notebook, l'objectif est de créer des outils permettant de générer un jeu de données initial pour d'entrainement d'algorithmes tels que les algorithmes de type bandits.

Le contexte est le suivant: générer des séries temporelles représentant les temps d'utilisation / de visionnage d'un utilisateur sur un site / une application de contenu long (ex. Youtube, Netflix, Prime Video, ...). Chaque valeur de ces séries temporelles représentera alors le temps de visionnage de l'utilisateur sur une journée. 

Pour ce faire, on souhaite générer des profiles variés sur les propriétés suivantes :
- temps total de visionnage (longueur de la série temporelle)
- saisonnalité du temps de visionnage
- variation (variance) du temps de visionnage

Pour la saisonnalité, on peut penser à un signal périodique. Le plus simple, reste encore un signal sinusoidal:
- la saisonnalité pouvant être très forte ou très faible, cela est lié à l'amplitude du signal.
- la vitesse de changement de cette saisonnalité peut être rattachée à la fréquence du signal.

On pourra aussi additionner les signaux pour obtenir des saisonnalités plus complexes, avec des forte périodes creuses elles mêmes saisonnières.

Le temps total de visionnage peut être vu comme la longueur observée du signal.

En pratique, les signaux ne pourront être que discrets, donc un échantillonnage sera nécessaire.
C'est sur cet échantionnage qu'on pourra alors modifier la variation du signal. On pensera par exemple à une loi normale générant du bruit avec une certaine variance et qu'on additionnera à notre échantillon, ce qui reviendra à bruiter l'échantillon avec une variation controlée.

Une autre notion liée à la variation est l'espacement de ces fortes variations, qui peut être vue comme une "dilution" du bruit, permettant ainsi d'avoir des plages de temps assez faibles et d'autres bien plus fortes et de contrôler la taille de celles-ci.
Cette notion de dilution du bruit du signal peut être représentée par une loi binomiale qui à chaque valeur de notre échantillon choisi avec une probabilité donnée `p` si le bruit généré pour cette valeur doit être appliqué ou non, voire même avec quel pourcentage.
C'est donc un bruit supplémentaire, mais qui permet une maîtrise plus précise des variations du signal.

On aura l'occasion de tester et illustrer l'impact de ces différents paramètres par la suite.

### Saisonnalité

Comme dit précédemment, on utilisera donc un signal sinusoidal, dont voici la défninition générale paramétrée par la taille observée du signal, sa fréquence et la fréquence d'échantionnage voulue:

```python
def idealSeasonalTimeSeries(size, freq, fe):
    '''
    Generate a perfect sinusoidal signal with a given @freq and returns a sample of a given @size and a sampling frequency @fe.
    '''
    x = np.arange(0, size, 1/fe)
    return np.cos(2*np.pi * freq * x)
```

NB: on pourra gérer la force de la saisonnalité (entre autre la variation à long-terme) grâce à l'amplitude du signal. Ceci correspond à une constante multiplicative `A` devant le signal.

### Variations

```python
def normalJitter(data, factor, dilution=0, randomstate=None, randgen=None):
    '''
    Generate a normal-noise and add it to given @data.
    '''
    if factor == 0 or dilution == 1:
        return np.array(data)
    
    randgen = np.random.default_rng(randomstate) if randgen is None else randgen

    size = len(data)
    
    noise = randgen.normal(0, factor, size)
    dilution_noise = randgen.binomial(1, 1-dilution, size) if dilution > 0 else 1
    
    return np.array(data) + (dilution_noise) * np.array(noise)
```

### Tests & illustrations

```python
def drawTestJittering(
    size=3,
    freq=1,
    fe=None,
    factor=0.25,
    dilution=0,
    
    randomstate=None,
    title=None,
    titlesize=12
):
    fe = freq*100 if fe is None else fe

    # sampling abscisse values
    x = np.arange(0, size, 1/fe)
    # ideal signal sampling values
    y = idealSeasonalTimeSeries(size, freq, fe)
    # noisy signal sampling values
    noisy_y = normalJitter(y, factor, dilution=dilution, randomstate=randomstate)

    # draw resulting signal
    plt.figure(figsize=(20,7))
    
    plt.plot(x, y, 'o', color='blue')
    plt.plot(x, y, color='cyan')
    plt.plot(x, noisy_y, color='red')

    if(not (title is None)):
        plt.title(title, {'fontsize': titlesize})
    
    plt.show()
```

#### test size:

```python
for s in [1, 2, 5, 10]:
    drawTestJittering(
        size=s,
        randomstate=42,
        title=f"Test - Sample size of {s}"
    )
```

#### test frequency:

```python
for f in [0.5, 1, 5]:
    drawTestJittering(
        freq=f,
        randomstate=42,
        title=f"Test - frequency of {f}"
    )
```

#### test sampling frequency:

```python
for fe in [1, 1*10, 1*100, 1*1000]:
    drawTestJittering(
        freq=1,
        fe=fe,
        randomstate=42,
        title=f"Test - sampling frequency of {fe} (with a given freq of 1)"
    )
```

#### test noise factor:

```python
for f in [0, 0.25, 0.5, 0.75, 1]:
    drawTestJittering(
        factor=f,
        randomstate=42,
        title=f"Test - noise factor of {f*100}%"
    )
```

#### test noise dilution:

```python
for d in [0, 0.25, 0.5, 0.75, 1]:
    drawTestJittering(
        dilution=d,
        randomstate=42,
        title=f"Test - noise dilution of {d*100}%"
    )
```
