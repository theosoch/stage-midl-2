```python
def print_matrix(array):
    matrix = ''
    for row in array:
        try:
            for number in row:
                matrix += f'{number}&'
        except TypeError:
            matrix += f'{row}&'
        matrix = matrix[:-1] + r'\\'
    display(Math(r'\begin{bmatrix}'+matrix+r'\end{bmatrix}'))
```

```python
def GenerateMatrixS(K):
  S = np.zeros((2, K))
  return S
```

```python
def GenerateVisitorReward(arm_average, nums_rows):
  visitor_reward = np.random.normal(arm_average,0.5, size=(nums_rows, len(arm_average)))
  names = ['visitor reward for arm ' + str(i+1) for i in range(len(arm_average))]
  visitor_reward = pd.DataFrame(visitor_reward)
  visitor_reward.columns = names
  return visitor_reward
```

```python
def GenerateVisitorRewardForTS(arm_average, nums_rows):
  tableau = []
  for _ in range(nums_rows):
      ligne = [random.choices([0, 1], weights=[1-p, p])[0] for p in arm_average]
      tableau.append(ligne)
  tableau = pd.DataFrame(tableau)
  return tableau
```

```python
def PlayArm(iter, arm, S, visitor_reward):
    # Update mean
    S[0, arm] = ((S[0, arm] * S[1, arm] + visitor_reward.iloc[iter, arm]) / (S[1, arm] + 1))
    # Update play count
    S[1, arm] += 1
    return S
```

```python
def DataControlK(visitor_reward , K):
  if visitor_reward.shape[1] <2:
    raise ValueError("number of arms must be superior or equal to 2")
  if visitor_reward.shape[1] != K:
    raise ValueError("Each arm need a result")
```

```python
def DataControlContextReward(dt , visitor_reward):
  if dt.shape[0] != visitor_reward.shape[0]:
    raise ValueError("Number of row in contextual data and rewards data are not equals")
```

```python
def PlotEvolution(rewindlist , choicelist ,  method):
  namelist = ["arm" + str(i+1) for i in range(rewindlist.shape[1])]
  if method == 'epsilon':
    rewindlist.plot(label = namelist)
    plt.legend()
    plt.title('Evolution of estimated theta for Epsilon-Greedy Algorithm')
    plt.show()
    compte = [choicelist.count(i) for i in range(rewindlist.shape[1])]
    plt.figure(figsize = (6,4))
    plt.bar(namelist , compte)
    plt.title('Pick Rate of each arm')
    plt.show()
  if method == "ucb":
    rewindlist.plot(label = namelist)
    plt.legend()
    plt.title('Evolution of estimated theta for UCB Algorithm')
    plt.show()
    plt.figure(figsize = (6,4))
    compte = [choicelist.count(i) for i in range(rewindlist.shape[1])]
    plt.bar(namelist , compte)
    plt.title('Pick Rate of each arm')
    plt.show()
  if method == 'ts':
    rewindlist.plot(label = namelist)
    plt.legend()
    plt.title('Evolution of estimated theta for Thompson-Sampling Algorithm')
    plt.show()
    plt.figure(figsize = (6,4))
    compte = [choicelist.count(i) for i in range(rewindlist.shape[1])]
    plt.bar(namelist , compte)
    plt.title('Pick Rate of each arm')
    plt.show()
```

### ɛ-Greedy

```python
def ConditionForEpsilonGreedy(S , epsilon , K):
  res = np.random.choice([1, 2], size=1, replace=True, p=[1 - epsilon, epsilon])

  if res == 1:
    return np.argmax(S[0])

  else:
    return np.random.choice(np.delete(np.arange(K), np.argmax(S[0])), size=1).item()
```

```python
def EpsilonGreedy(visitor_reward , K , epsilon , auto):
  choice = []
  S = GenerateMatrixS(K)
  rewindlist = [[0]*K]
  epsilon_decrease = epsilon/(visitor_reward.shape[0]-3)

  for i in range(K):
    S = PlayArm(iter = i , arm = i , S = S , visitor_reward = visitor_reward)
    choice.append(i)
    rewindlist.append(S[0].copy())


  for j in range(K ,  visitor_reward.shape[0]):
    choice.append(ConditionForEpsilonGreedy(S , epsilon , K))
    S = PlayArm(j , choice[j] , S , visitor_reward )
    rewindlist.append(S[0].copy())
    #epsilon = max(epsilon - epsilon_decrease , 0)

  theta_hat = S[0]
  names = ['Estimated theta for arm ' + str(i+1) for i in range(K)]
  rewindlist = pd.DataFrame(rewindlist)
  rewindlist.columns = names
  if auto == True:
    return S , choice , theta_hat  , rewindlist
  elif auto == False:
    return S , choice , theta_hat
  else:
    return None
```

### UCB-alpha



```python
def ProbaMaxForUCB(S, iter, alpha, K):
    values = np.zeros(K)
    for j in range(K):
        values[j] = S[0, j] + alpha * np.sqrt((np.log(iter)) / S[1, j])
    return values
```

```python
def ConditionForUCB(S, iter, alpha, K):
  return np.argmax(ProbaMaxForUCB(S, iter, alpha, K))
```

```python
def UCB(visitor_reward , K , alpha , auto):
  choice = []
  proba = []
  S = GenerateMatrixS(K)
  rewindlist = [[0]*K]

  for i in range(K):
    proba.append(1/K)
    S = PlayArm(i, i, S,visitor_reward)
    choice.append((i))
    rewindlist.append(S[0].copy())
  for j in range(K, visitor_reward.shape[0]):
    choice.append(ConditionForUCB(S,j,alpha, K))
    proba.append(max(ProbaMaxForUCB(S, j, alpha, K)))
    S = PlayArm(j, choice[j], S, visitor_reward)
    rewindlist.append(S[0].copy())
  names = ['Estimated theta for arm ' + str(i+1) for i in range(K)]
  rewindlist = pd.DataFrame(rewindlist)
  rewindlist.columns = names
  theta_hat = S[0]

  if auto == True:
    return S , choice , theta_hat  , rewindlist
  elif auto == False:
    return S , choice , theta_hat
  else:
    return None
```

### Thompson-Sampling

```python
def ConditionForThompsonSampling(S , alpha , beta):
  values = []
  for j in range(S.shape[1]):
    values.append(np.random.beta( alpha + S[0][j] * S[1][j] , beta + S[1][j] - S[0][j]*S[1][j]))
  choice = np.argmax(values)
  proba = np.max(values)
  return choice , proba
```

```python
def ThompsonSampling(visitor_reward , K , alpha , beta , auto):
  choice = []
  proba = []
  S = GenerateMatrixS(K)
  rewindlist = [[0]*K]

  for j in range(K):
    choice.append(j)
    proba.append(1/K)
    S = PlayArm(j, j, S, visitor_reward)
    rewindlist.append(S[0].copy())

  for j in range(K , visitor_reward.shape[0]):
    choice_j , proba_j  = ConditionForThompsonSampling(S , alpha , beta )
    choice.append(choice_j)
    proba.append(proba_j)
    S = PlayArm(j , choice[j] , S , visitor_reward)
    rewindlist.append(S[0].copy())

  names = ['arm' + str(i+1) for i in range(K)]
  rewindlist = pd.DataFrame(rewindlist)
  rewindlist.columns = names
  theta_hat = S[0]
  if auto == True:
    return S , choice , theta_hat  , rewindlist
  elif auto == False:
    return S , choice , theta_hat
  else:
    return None
```

### Regret

```python
def CumulativeRegret(choice, visitor_reward):
    cumulative_regret = []
    global_average = np.mean(visitor_reward, axis=0)
    max_average = np.max(global_average)
    cumulative_sum = 0
    for i in range(0, len(choice)):
      cumulative_sum += (max_average - global_average[choice[i]])
      cumulative_regret.append(cumulative_sum)
    return cumulative_regret
```

```python
def PlotRegret(regret_list):
  plt.plot(range(len(regret_list)), regret_list)
  plt.xlabel('Iteration')
  plt.ylabel('Cumulative Regret')
  plt.title('Cumulative Regret over Iterations')
  plt.show()
```

```python
def SimpleRegret(choice , visitor_reward):
  regret = []
  for i in range(visitor_reward.shape[0]):
    regret.append(RegretValue((choice[i]) , visitor_reward.iloc[i]))
  return regret
```

```python
def RegretValue(arm , vec_visitor_reward):
  return float(vec_visitor_reward.max() - vec_visitor_reward[arm])
```

### LINUCB

on va avoir une amtrice qui va stocker a chaque fois la covariance de nos variables , donc si on regarde nos parametres theta_hat qui va avoir la taille n features qui est la dimension du contexte , le theta hat il va avoir une forme , le theta hat est une amtrice k par n , ca va faire un produit matriciel puis nupperbhond , dans la regression , on va avoir aX + b , b va etre le beta donc la matrice de coefficient , elle est comme theta_hat n par nf , on va avoir une matrice qui va tout sauvegarder pour chaque bras , c'est ce qui sera dans A , ca sera une matrice de covariance en gros , , d'abord on va diagonaliser A (matrice de covariance associe a un bras K = 1) , objectif 1ere boucle , avoir K matrice diagonalisee. On appelle notre contexte qu'on stocke dans xi , pour la regression , on fait l'inverse de la matrice de covariance A , th_hat = produit matricial de A par notre vecteur petit b , on va vouloir en,suite la prediction ta .on va creer ensuite nos upper bound et notre moyenne , on multiplie par notre bonus alpha pour recompenser notre modele.  On met ensuite a jour nos matrices ainsi que nos choix et proba et rewards. b recupere les recompenses et A fais le contexte et on renvoie le tout. A represnete la covariance et AX donc 1 dimension de plus

```python
def LINUCB(dt, visitor_reward, alpha, K):

  DataControlK(visitor_reward , K)
  DataControlContextReward(dt , visitor_reward)

  D = np.array(dt)
  visitor_reward = np.array(visitor_reward)
  n , m = D.shape
  choices = [0]*n
  rewards = [0]*n
  proba = [0]*n

  th_hat = np.zeros((K,m))
  b = np.zeros((K , m))
  A = np.zeros((K,m,m))

  p = [0]*K

  for j in range(K):
    A[j]=np.eye(m)
  for i in range(K):
    x_i = np.array(D[i])
    for j in range(K):
      A_inv = np.linalg.inv(np.transpose(A[j]))
      th_hat[j] = np.dot(A_inv , b[j])
      ta = np.transpose(x_i) @ A_inv @x_i
      upper_ci = alpha * np.sqrt(ta)
      mean_j = th_hat[j]   @x_i
      p[j] = upper_ci + mean_j
    choices[i] =i
    proba[i]=p[i]
    rewards[i]= visitor_reward[i , choices[i]]
    A[choices[i]] += np.outer(x_i , np.transpose(x_i))
    b[choices[i]] += x_i *rewards[i]

  for i in range(K , n):
    x_i = np.array(D[i])
    for j in range(K):
      A_inv = np.linalg.inv(np.transpose(A[j]))
      th_hat[j] = np.dot(A_inv , b[j])
      ta = np.transpose(x_i) @ A_inv @x_i
      upper_ci = alpha * np.sqrt(ta)
      mean_j = th_hat[j] @ x_i
      p[j] = upper_ci + mean_j

    choices[i] = np.argmax(p)
    proba[i]= max(p)
    rewards[i]= visitor_reward[i , choices[i]]
    A[choices[i]] += np.outer(x_i , np.transpose(x_i))
    b[choices[i]] += x_i *rewards[i]

  return A , b , th_hat , choices
```

```python
def EpsilonGreedyMultiPlot(visitor_reward , K , epsilon_list ):
  plt.figure(figsize = (10 , 8))
  for epsilons in epsilon_list:
    X = EpsilonGreedy(visitor_reward , K , epsilons , False)[1]
    Y = CumulativeRegret(X , visitor_reward)
    plt.plot( range(len(Y)) , Y , label = r'$\varepsilon$ = ' + str(epsilons))
  plt.title( r' Cumulative regret according to $\varepsilon$ for $\varepsilon$ - Greedy algorithm')
  plt.legend()
  plt.show()
  return None
```

```python
def UCBMultiPlot(visitor_reward , K , alpha_list ):
  for alphas in alpha_list:
    X = UCB(visitor_reward , K , alphas , False)[1]
    Y = CumulativeRegret(X , visitor_reward)
    plt.plot( range(len(Y)) , Y , label = r'$\alpha$ = ' + str(alphas))
  plt.title( r' Cumulative regret according to $\alpha$ for UCB-$\alpha$ algorithm')
  plt.legend()
  plt.show()
  return None
```

```python
def TSMultiPlot(visitor_reward , K , alpha , beta , N):
  for i in range(N):
    X = ThompsonSampling(visitor_reward , K , alpha , beta , False)[1]
    Y = CumulativeRegret(X , visitor_reward)
    plt.plot( range(len(Y)) , Y)
  plt.title( r' Cumulative regret for Thompson-Sampling algorithm for N générations')
  plt.legend()
  plt.show()
  return None
```

### Creating GameBoard

```python
def CreateGameBoard(lines , columns):
  GB = []
  for i in range(lines):
    GB.append([0] * columns)
  return GB
```

```python
def DefineGoal(GameBoard , goal_x , goal_y , val_goal):
  GameBoard[goal_x][goal_y] = val_goal
  coord = (goal_x , goal_y)
  return GameBoard  , coord
```

```python
def Getcases_actions(GB):
  cases = []
  for i in range(len(GB)):
    for j in range(len(GB[0])):
      cases.append((i,j))
  actions = ["up" , "right" , "down" , "left"]
  return cases , actions
```

```python
def DefineLake(GB , lake_x , lake_y , loss_value):
  GB[lake_x][lake_y] = loss_value
```

```python
def GetSpawnLooseEndcases(GB , goal_coordinate):
  spawncases = []
  endingcases = []
  for i in range(len(GB)):
    for j in range(len(GB[0])):
      if GB[i][j] == 0:
        spawncases.append((i,j))
      else:
        endingcases.append((i,j))
  losingcases= endingcases.copy()
  losingcases.remove(goal_coordinate)
  return spawncases , losingcases , endingcases
```

```python
def CreateSpawnPoint(GB , random_spawn):
  if random_spawn == False:
    X = int(input("Select line spawn point"))
    if X < 0 or X > len(GB):
      raise ValueError("this spawn point coordinate isn't in the GameBoard selected")
    Y = int(input("Select row spawn point"))
    if X < 0 or X > len(GB[0]):
      raise ValueError("this spawn point coordinate isn't in the GameBoard selected")
    spawnpoint = (X,Y)
    return spawnpoint
```

### IA function

```python
def Movement(GB , current_case , direction):
  i = current_case[0]
  j = current_case[1]
  max_height = len(GB)-1
  max_lenght = len(GB[0])-1
  if direction == "up":
    return (max(0, i-1) , j)
  if direction =="right":
    return (i , min(max_lenght , j+1))
  if direction == "down":
    return (min(max_height , i+1) , j)
  if direction =="left":
    return(i , max(0 , j-1))
```

```python
def GetReward(GB , current_case):
  return GB[current_case[0]][current_case[1]]
```

```python
def BellmanFormula(Q , current_case , movement , next_case ,reward_gained , alpha , gamma):
  return Q.loc[current_case , movement] + alpha * (reward_gained + gamma * max( Q.loc[next_case].values) - Q.loc[current_case ,movement])
```

```python
def PlotcaseEvolution(casecountlist , winlooselist):
  plt.figure(figsize = (16,8))
  plt.plot(casecountlist)
```

```python
def FrozenLakeInitiate(GB , alpha , gamma , epsilon , N , random_spawn , goal_coordinate):
  #condition
  ConditionForParameter(alpha)
  ConditionForParameter(gamma)
  ConditionForParameter(epsilon)
  #SpecialCases
  cases_with_spawn , losing_cases , ending_case = GetSpawnLooseEndcases(GB , goal_coordinate)

  # variables initialisation
  cases , actions = Getcases_actions(GB)
  Q = InitialisationQvalues(cases , actions)
  result = {}
  winloose = []
  epsilon_loose = epsilon/N
  casecountlist = []

  # spawnpoint choice
  spawnpoint = CreateSpawnPoint(GB , random_spawn)

  # Training
  starting_time = time.time()
  for generation in range(1,N+1):
    case_count =0
    winloose.append('Win')
    Printalgotime(starting_time , N , generation)

    #Spawnpoint
    if random_spawn == True:
      spawnpoint = random.choice(cases_with_spawn)
    else:
      spawnpoint = spawnpoint
    current_case = spawnpoint

    #Processus
    while(current_case not in ending_case):
      choice = np.random.uniform(0,1)
      if choice > epsilon:
        if (Q.loc[current_case]==0).all():
          movement = random.choice(actions)
        else:
          movement = Q.loc[current_case].idxmax()
      else:
        movement = random.choice(actions)

      next_case = Movement(GB , current_case , movement)
      reward_gained = GetReward(GB , next_case)
      Q.loc[current_case,movement] = BellmanFormula(Q , current_case , movement , next_case , reward_gained , alpha , gamma)
      case_count +=1
      current_case = next_case

      if (current_case in losing_cases):
        winloose[-1] = 'Game Over'
    epsilon = max(epsilon - epsilon_loose,0)
    casecountlist.append(case_count)

  for index , row in Q.iterrows():
    x =  index
    best_action = max(actions , key = lambda b : row[b])
    result[x] =  best_action
  new_Q = pd.DataFrame(Q)
  result = pd.DataFrame(list(result.items()) , columns = ['position' , 'recommanded action'])
  time4 = time.time()
  final_time = time4 - starting_time
  print("Training done at 100% , total time:" , round(final_time,3) , ' seconds')
  print("Succes Rate: " , winloose.count("Win") * 100/N, '%' )

  return result , new_Q , winloose , cases_with_spawn , losing_cases , ending_case , casecountlist
```

```python
def ModelEvaluation(GB , Q , N , spawnlist , loosinglist , endinglist , random_spawn):
  winloose = []

  if random_spawn == False:
    X = int(input("Select line spawn point "))
    if X < 0 or X > len(GB):
      raise ValueError("this spawn point coordinate isn't in the GameBoard selected")
    Y = int(input("Select row spawn point "))
    if X < 0 or X > len(GB[0]):
      raise ValueError("this spawn point coordinate isn't in the GameBoard selected")
    spawnpoint = (X,Y)

  for generation in range(N): #initialisation
    if generation == int(N/4):
      print('Evaluation done at 25%')
    if generation == int(N/2):
      print('Evaluation done at 50%')
    if generation == int(3*N/4):
      print('Evaluation done at 75%')

    winloose.append("Win")
    #spawning
    cases_count = 0
    if random_spawn == True:
      spawnpoint = random.choice(spawnlist)
    current_case = spawnpoint

    while current_case not in endinglist:
      movement = Q.loc[current_case].idxmax()
      next_case = Movement(GB , current_case , movement)
      if next_case in loosinglist:
        winloose[-1] = "Game Over"
      current_case = next_case
  print("Evaluation Complete")
  print("IA Succes Rate: " , winloose.count("Win") * 100/N , '%')
```

```python
def PlotHeatMap(Qtable , resume , loosinglist):
  x,y = [val +1 for val in result.values[-1][0]]

  token = 0
  whole= []
  f , ax = plt.subplots(figsize = (11,9))
  #on va reshape le DataFrame
  for i in range(x):
    Yline = []
    for j in range(y):
      Yline.append(Qtable.iloc[token].max())
      token +=1
    whole.append(Yline)
  cmap = sns.diverging_palette(220, 10, as_cmap=True)
  sns.heatmap(whole ,cmap = cmap , vmax = 1.1 , square = True , xticklabels = x +1 , yticklabels = y +1 , linewidths=.5, cbar_kws={"shrink": .5}, ax=ax, annot=True, fmt="f")

  token2 = 0
  for i in range(x):
    for j in range(y):
      if result.iloc[token2]['recommanded action'] == 'up':
        plt.text(j+0.5, i+0.7, u'\u2191', fontsize=12)
      elif result.iloc[token2]['recommanded action'] == 'down':
        plt.text(j+0.5, i+0.7, u'\u2193', fontsize=12)
      elif result.iloc[token2]['recommanded action'] == 'left':
        plt.text(j+0.5, i+0.7, u'\u2190', fontsize=12)
      elif result.iloc[token2]['recommanded action'] == 'right':
        plt.text(j+0.5, i+0.7, u'\u2192', fontsize=12)

      if (i,j) in loosinglist:
        plt.text(j + 0.5  , i + 0.3 ,'Lake' , fontsize = 14 , ha = 'center' , va ='center' , color = 'navy')
      token2 +=1

  plt.text(x -0.5  , y - 0.7 ,'Goal' , fontsize = 14 , ha = 'center' , va ='center' , color = 'white')
  plt.title('Heatmap of trained QTable with directions and quality of the action')
```
