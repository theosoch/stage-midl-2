```python
import numpy as np

# Create two sequences
s1 = [3, 1, 2, 2, 1]
s2 = [2, 0, 0, 3, 3, 1, 0]
```

```python

s1 = [5, 7 ,6 , 4 , 4 , 2 , 7 ,7]
s2 = [7 , 2 , 4 , 8 , 4 , 4 , 1]
```

```python

n, m = len(s1), len(s2)
dtw_matrix = np.zeros((n, m))

for i in range(n):
    for j in range(m):
        dtw_matrix[i, j] = abs(s1[i] - s2[j])**2

dtw_distance = dtw_matrix[0, 0]

print(dtw_matrix)


#init first line
for i in range(1,n):
  dtw_matrix[i, 0]= dtw_matrix[i, 0] + dtw_matrix[i-1, 0]

for j in range(1,m):
  dtw_matrix[0, j]= dtw_matrix[0, j] + dtw_matrix[0, j-1]


for i in range(1,n):
  for j in range(1,m):
    dtw_matrix[i, j]= dtw_matrix[i, j] +  np.min([dtw_matrix[i-1, j],dtw_matrix[i, j-1],dtw_matrix[i-1, j-1]])

print(dtw_matrix)
print(dtw_matrix[i, j])
```

```python
import pandas as pd
import numpy as np# Plotting Packages
import matplotlib.pyplot as plt
import seaborn as sbn# Configuring Matplotlib
import matplotlib as mpl
mpl.rcParams['figure.dpi'] = 300
savefig_options = dict(format="png", dpi=300, bbox_inches="tight")# Computation packages
from scipy.spatial.distance import euclidean
#from fastdtw import fastdtw
```

```python
def compute_euclidean_distance_matrix(x, y) -> np.array:
    """Calculate distance matrix
    This method calcualtes the pairwise Euclidean distance between two sequences.
    The sequences can have different lengths.
    """
    dist = np.zeros((len(y), len(x)))
    for i in range(len(y)):
        for j in range(len(x)):
            dist[i,j] = (x[j]-y[i])**2
    return dist
```

```python
def compute_accumulated_cost_matrix(x, y) -> np.array:
    """Compute accumulated cost matrix for warp path using Euclidean distance
    """
    distances = compute_euclidean_distance_matrix(x, y)

    # Initialization
    cost = np.zeros((len(y), len(x)))
    cost[0,0] = distances[0,0]

    for i in range(1, len(y)):
        cost[i, 0] = distances[i, 0] + cost[i-1, 0]

    for j in range(1, len(x)):
        cost[0, j] = distances[0, j] + cost[0, j-1]

    # Accumulated warp path cost
    for i in range(1, len(y)):
        for j in range(1, len(x)):
            cost[i, j] = min(
                cost[i-1, j],    # insertion
                cost[i, j-1],    # deletion
                cost[i-1, j-1]   # match
            ) + distances[i, j]

    return cost
```

```python
# Create two sequences
x = [3, 1, 2, 2, 1]
y = [2, 0, 0, 3, 3, 1, 0]
```

```python

x = [5, 7 ,6 , 4 , 4 , 2 , 7 ,7]
y = [7 , 2 , 4 , 8 , 4 , 4 , 1]
```

```python
cost_matrix = compute_accumulated_cost_matrix(x, y)
```
